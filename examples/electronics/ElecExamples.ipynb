```python
## h = 1e-6s
ref = np.loadtxt("DiodeBridgePowSup/DiodeBridgePowSupTRAP1us.dat")
refspice = np.loadtxt("DiodeBridgePowSup/SPICEN0p25TRAP1us.dat", skiprows=8)

## h = 1e-7s
#ref = np.loadtxt("DiodeBridgePowSup/DiodeBridgePowSupTRAP0p1us.dat")
#refspice = np.loadtxt("DiodeBridgePowSup/SPICEN0p25TRAP0p1us.dat", skiprows=8)



mine = np.loadtxt("DiodeBridgePowSup/DiodeBridgePowSup.dat")
nbrows = np.min([mine.shape[0], refspice.shape[0]])
nbrows_ref = ref.shape[0]

col = 1 # Supply voltage 
plt.figure(col, figsize=(20,5))
plt.subplot(1,3,1)
plt.plot(ref[:,0], ref[:,col], 'go')
plt.plot(mine[:,0], mine[:,col], 'r--')
plt.plot(refspice[:,0], refspice[:,col], 'b--')
plt.xlim(left=0.3e-6)

plt.title("Supply voltage")
plt.subplot(1,3,2)
plt.plot(ref[:,0], ref[:,col] - mine[:nbrows_ref,col], '+-')
plt.subplot(1,3,3)
plt.plot(refspice[:nbrows,0], refspice[:nbrows,col] - mine[:nbrows,col], '+-')
plt.xlim(left=0.3e-6)

col = 3 # diode R1 current
plt.figure(col, figsize=(20,5))
plt.subplot(1,3,1)
plt.plot(ref[:,0], ref[:,col], 'go')
plt.plot(mine[:,0], mine[:,col], 'r--')
plt.plot(refspice[:,0], refspice[:,6], 'b--')
plt.xlim(left=0.3e-6)
plt.title("diode R1 current")
plt.subplot(1,3,2)
plt.plot(ref[:,0], ref[:,col] - mine[:nbrows_ref,col], '+-')
plt.subplot(1,3,3)
plt.plot(refspice[:nbrows,0], refspice[:nbrows,6] - mine[:nbrows,col], '+-')
plt.xlim(left=0.3e-6)

col = 6 # diode F1 current
plt.figure(col, figsize=(20,5))
plt.subplot(1,3,1)
#plt.plot(ref[:,0], ref[:,col], 'go')
plt.plot(mine[:,0], mine[:,col], 'r--')
plt.plot(refspice[:,0], refspice[:,5], 'b--')
plt.ylim(0.,0.04)
plt.xlim(left=0.3e-6)
plt.title("diode F1 current")
plt.subplot(1,3,2)
plt.plot(ref[:,0], ref[:,col] - mine[:nbrows_ref,col], '+-')
plt.subplot(1,3,3)
plt.plot(refspice[:nbrows,0], refspice[:nbrows,5] - mine[:nbrows,col], '+-')
plt.xlim(left=0.3e-6)

col = 4 # diode F1 current
plt.figure(col, figsize=(20,5))
plt.subplot(1,3,1)
plt.plot(ref[:,0], -(ref[:,col] + ref[:,col +1]), 'go')
plt.plot(mine[:,0], -(mine[:,col] + mine[:,col+1]), 'r--')
plt.plot(refspice[:,0], refspice[:,3], 'b--')
plt.xlim(left=0.3e-6)
plt.title("resistor voltage")
plt.subplot(1,3,2)
plt.plot(ref[:,0], ref[:,col] - mine[:nbrows_ref,col], '+-')
plt.subplot(1,3,3)
plt.plot(refspice[:nbrows,0], refspice[:nbrows,3] + (mine[:nbrows,col] + mine[:nbrows,col+1]), '+-')
plt.xlim(left=0.3e-6)
```
